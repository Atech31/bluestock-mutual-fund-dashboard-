```python
import numpy as np
import pandas as pd

# 1. Generate Synthetic FinTech Dataset
np.random.seed(42)
n_rows = 1000

data = {
    "Transaction_ID": [f"TXN_{1000 + i}" for i in range(n_rows)],
    "Customer_ID": [f"CUST_{np.random.randint(100, 200)}" for _ in range(n_rows)],
    "Order_Date": pd.date_range(
        start="2026-01-01", periods=n_rows, freq="h"
    ).strftime("%Y-%m-%d"),
    "Product_Category": np.random.choice(
        ["Crypto Trade", "Peer Transfer", "Micro Loan", "Wealth Management"],
        size=n_rows,
    ),
    "Transaction_Amount": np.random.choice(
        [50.0, 150.0, 300.0, 1200.0, np.nan, 5000.0], size=n_rows, p=[0.3, 0.3, 0.2, 0.1, 0.05, 0.05]
    ),
    "Status": np.random.choice(
        ["Completed", "Failed", "Refunded"], size=n_rows, p=[0.8, 0.1, 0.1]
    ),
    "Region": np.random.choice(
        ["North America", "Europe", "Asia-Pacific", "LATAM"], size=n_rows
    ),
}

df = pd.DataFrame(data)

# Inject duplicates to test cleaning logic
df = pd.concat([df, df.iloc[:15]], ignore_index=True)

# 2. Data Cleaning Routine
df.drop_duplicates(inplace=True)
median_amt = df["Transaction_Amount"].median()
df["Transaction_Amount"].fillna(median_amt, inplace=True)

# 3. Export Cleaned Dataset
df.to_csv("fintech_transactions_cleaned.csv", index=False)
print("✅ 'fintech_transactions_cleaned.csv' successfully created!")
print(df.info())
```
